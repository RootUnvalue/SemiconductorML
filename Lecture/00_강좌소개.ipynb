# 00_강좌소개

## 반도체머신러닝 강좌 소개

본 강좌는 인공지능 분야의 핵심 요소인 머신러닝에 대한 체계적인 이해를 목표로 합니다. 데이터 기반의 문제 해결 능력을 함양하고, 실제 현장에서 적용 가능한 머신러닝 기술을 습득할 수 있도록 이론과 실습을 병행하여 구성되었습니다. 현대 사회의 다양한 데이터를 분석하고 예측하는 데 필수적인 머신러닝의 원리와 응용 방법을 심도 있게 다룹니다.

### 강좌 목표
*   머신러닝의 기본 개념 및 핵심 이론을 명확히 이해합니다.
*   선형 회귀, 로지스틱 회귀, 서포트 벡터 머신(SVM), 결정 트리, 앙상블 모델 등 주요 머신러닝 알고리즘의 동작 원리를 학습합니다.
*   Python 프로그래밍 언어와 NumPy, Pandas, Scikit-learn, Matplotlib, Seaborn 등 필수 라이브러리를 활용하여 데이터 전처리, 모델 구축 및 평가 과정을 실습합니다.
*   실제 데이터셋을 분석하고 머신러닝 모델을 적용하여 당면한 문제에 대한 해결책을 도출하는 능력을 배양합니다.

### 강좌 대상
*   머신러닝 분야에 대한 기초 지식 습득을 희망하는 학습자
*   데이터 분석가, 데이터 과학자, 머신러닝 엔지니어 지망생
*   데이터 기반의 의사결정 역량 강화를 필요로 하는 실무자

### 선수 과목 및 준비 사항
*   Python 프로그래밍 기초 문법에 대한 이해
*   선형대수학 및 통계학의 기본적인 개념 이해 (필수 아님, 그러나 학습에 도움이 될 수 있음)
*   본 강좌는 Google Colaboratory 환경을 기반으로 진행되므로 별도의 소프트웨어 설치는 요구되지 않습니다.

### 교재
*   으뜸 머신러닝 - 김영민, 박동규, 김성수 지음 (주) 생능출판사

### 강사 소개
*   이남용 교수, 인제대학교 AI소프트웨어학과, nylee@inje.ac.kr

### 성적 평가 방법

### 강좌 운영 방식
매주 월요일, 총 39개의 강의 중 3개씩 e-edu.inje.ac.kr을 통해 업로드될 예정입니다. 각 강의에는 해당 강의의 숙제가 함께 발표됩니다.

본 강좌를 통해 머신러닝의 전문 지식을 습득하고, 여러분의 학문적 및 실무적 역량을 한층 더 강화하시기를 기대합니다.

### Google Colab 사용 안내

본 강좌는 Google Colaboratory(Colab) 환경에서 진행됩니다. Colab은 클라우드 기반의 Jupyter 노트북 환경으로, 별도의 설정 없이 웹 브라우저를 통해 Python 코드를 작성하고 실행할 수 있습니다.

학생 여러분은 다음 사항을 준비해 주시기 바랍니다.

1.  **Google 계정:** Colab 사용을 위해서는 Google 계정(Gmail 주소)이 필수적입니다. 계정이 없는 경우 미리 생성해 주시기 바랍니다.
2.  **Colab 접속:** 웹 브라우저(Chrome 권장)를 통해 [colab.research.google.com](https://colab.research.google.com/)에 접속합니다.
3.  **노트북 열기:** 강의 자료로 제공되는 `.ipynb` 파일을 Colab에서 열거나, 새로 만들기를 통해 실습을 진행합니다.
4.  **런타임 설정:** GPU 사용이 필요한 경우, '런타임' 메뉴 -> '런타임 유형 변경'에서 '하드웨어 가속기'를 'GPU'로 설정할 수 있습니다.

Colab 사용에 어려움이 있는 경우, 언제든지 강사에게 문의해 주십시오.
