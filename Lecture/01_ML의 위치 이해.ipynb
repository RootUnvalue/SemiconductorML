## How We Solve Problems

이 강의의 대부분의 실습은 다음과 같은 과정을 따른다.

> **Problem → Problem Analysis → Method → Application → Result Analysis**

### Problem

가장 먼저 다음 두 가지를 생각한다.

- 무엇을 하고자 하는가?
- 어떤 Data를 가지고 있는가?

Machine Learning 문제는 Algorithm에서 시작하는 것이 아니라 **Problem에서 시작해야 한다.**

예를 들어 다음과 같이 시작하지 않는다.

> "Linear Regression을 사용해 보자."

대신 다음과 같은 Problem에서 시작한다.

> "주택에 대한 정보를 이용하여 주택 가격을 예측할 수 있을까?"

어떤 Machine Learning Method를 사용할 것인지는 Problem을 이해한 다음 결정한다.


### Problem Analysis

Problem과 주어진 Data를 분석한다.

다음과 같은 질문을 생각할 수 있다.

- 어떤 정보가 주어져 있는가?
- 어떤 정보가 중요한가?
- 무엇을 예측하거나 발견하려고 하는가?
- 어떤 형태의 결과가 필요한가?


### Method

Problem을 해결하기에 적절한 Method를 선택한다.

Problem에 따라 서로 다른 Machine Learning Method가 필요할 수 있다.

예를 들면 다음과 같다.

- Linear Regression
- Logistic Regression
- k-Nearest Neighbors
- Decision Tree
- Random Forest
- Support Vector Machine
- K-Means
- Artificial Neural Network

이 강의에서는 다양한 Machine Learning Method를 경험한다.

중요한 질문은 단순히

> "이 Method는 어떻게 동작하는가?"

가 아니다.

더 중요한 질문은

> **"이 Method를 어떤 Problem에 사용할 수 있는가?"**

이다.


### Application

선택한 Method를 실제 Data에 적용한다.

Python과 Machine Learning Library를 이용하면 비교적 적은 양의 Code로 다양한 Machine Learning Method를 구현할 수 있다.

AI Tool을 이용하여 필요한 Code를 생성하거나 수정할 수도 있다.

그러나 생성된 Code가 무엇을 수행하는지는 이해해야 한다.


### Result Analysis

결과를 얻었다고 해서 Data Analysis가 끝나는 것은 아니다.

다음과 같은 질문을 해야 한다.

- 이 Result는 무엇을 의미하는가?
- Result가 합리적인가?
- 얼마나 좋은 Result인가?
- 처음 제시한 Problem에 대한 답이 되는가?
- 다른 해석은 가능한가?

실제 문제에서는 동일한 Data를 분석하더라도 서로 다른 결론에 도달할 수 있다.

따라서 **Result를 얻는 것만큼 Result를 해석하는 것이 중요하다.**

## Three Types of Problems

과학, 공학 및 Data Analysis의 많은 문제는 Input과 Output 사이의 관계로 표현할 수 있다.

다음과 같은 관계가 있다고 하자.

$$
y=f(x)
$$

여기서

- $x$ : **Input**
- $y$ : **Output**
- $f$ : Input과 Output 사이의 **Relationship**

이다.

무엇을 알고 있고 무엇을 구하려고 하는가에 따라 Problem을 세 가지 형태로 생각할 수 있다.


### Forward Problem

$x$와 $f$를 알고 있고 $y$를 구하는 문제이다.

$$
x,\ f \rightarrow y
$$

즉,

$$
y=f(x)
$$

에서 알려진 Function $f$에 Input $x$를 적용하여 Output $y$를 계산한다.

예를 들어 Celsius와 Fahrenheit 사이의 관계를 알고 있다고 하자.

$$
F=\frac{9}{5}C+32
$$

$C=30$이 주어졌다면 이 식을 이용하여 $F$를 계산할 수 있다.

이것은 **Forward Problem**이다.

Rule이 이미 알려져 있으며, 알려진 Rule을 Input에 적용하여 Output을 구한다.


### Inverse Problem

$y$와 $f$를 알고 있고 $x$를 구하는 문제이다.

$$
y,\ f \rightarrow x
$$

예를 들어

$$
y=x^2
$$

이라는 관계를 알고 있고

$$
y=25
$$

가 주어졌다고 하자.

우리는 이 Output을 발생시킬 수 있는 Input $x$를 찾으려고 한다.

즉, Output으로부터 Input을 역으로 찾는다.

이것은 **Inverse Problem**이다.


### Learning Problem

이번에는 $x$와 $y$에 대한 관측 Data는 있지만 $f$를 모른다고 하자.

예를 들어 다음과 같은 주택 Data가 있다고 하자.

| Size | Price |
|---:|---:|
| 50 | 150 |
| 70 | 210 |
| 90 | 280 |
| 120 | 360 |

여러 개의 관측값

$$
(x_1,y_1),(x_2,y_2),\ldots,(x_n,y_n)
$$

은 알고 있지만,

$$
y=f(x)
$$

에서 $f$가 무엇인지는 알지 못한다.

우리의 목적은 주어진 Data를 이용하여 $f$를 근사하는 Model을 찾는 것이다.

$$
(x_1,y_1),(x_2,y_2),\ldots,(x_n,y_n)
\rightarrow \hat{f}
$$

학습된 Model을 얻으면 새로운 Input $x$에 대하여

$$
\hat{y}=\hat{f}(x)
$$

를 이용하여 Output을 예측할 수 있다.

이것이 **Learning Problem**이다.

Machine Learning은 주로 이러한 Learning Problem을 다룬다.


## What Is Machine Learning?

Traditional Programming에서는 사람이 Rule을 만들어 Computer에 알려준다.

이를 간단히 표현하면 다음과 같다.

> **Data + Rules → Results**

Machine Learning에서는 Data로부터 Rule이나 Relationship을 학습한다.

> **Data → Learning → Model**

학습된 Model은 새로운 Data에 적용할 수 있다.

> **New Data + Model → Prediction**

실제 관계가

$$
y=f(x)
$$

라고 하자.

Machine Learning에서는 Data를 이용하여 이를 근사하는 Model

$$
\hat{f}(x)\approx f(x)
$$

를 찾는다.

그리고 새로운 Input에 대하여

$$
\hat{y}=\hat{f}(x)
$$

를 이용하여 Prediction을 수행한다.

따라서 Machine Learning에서는

> "Rule을 어떻게 직접 Programming할 것인가?"

대신

> **"Data로부터 적절한 Rule을 학습할 수 있는가?"**

라는 질문을 하게 된다.

## Major Types of Machine Learning

Machine Learning은 학습하는 방식에 따라 크게 다음과 같이 구분할 수 있다.

- Supervised Learning
- Unsupervised Learning
- Reinforcement Learning


### Supervised Learning

**Supervised Learning**에서는 Input과 이에 대응하는 Output이 함께 주어진다.

Dataset을 다음과 같이 표현할 수 있다.

$$
D=\{(x_i,y_i)\}_{i=1}^{n}
$$

이 Data를 이용하여

$$
f:x\rightarrow y
$$

의 관계를 학습한다.

대표적인 Supervised Learning Problem은 다음과 같다.

- Regression
- Classification


#### Regression

**Regression**은 연속적인 Numerical Value를 예측하는 문제이다.

예를 들면 다음과 같다.

- House Price
- Temperature
- Sales
- Fuel Efficiency

예를 들어,

> **House Information → House Price**

와 같은 문제이다.


#### Classification

**Classification**은 Category를 예측하는 문제이다.

예를 들면 다음과 같다.

- Spam / Normal
- Disease / No Disease
- Dog / Cat
- Leave / Stay

예를 들어,

$$
\text{Email}
\rightarrow
\{\text{Spam},\text{Normal}\}
$$

과 같은 문제이다.


### Unsupervised Learning

**Unsupervised Learning**에서는 Input Data는 있지만 정답에 해당하는 Output은 주어지지 않는다.

Dataset은 다음과 같이 표현할 수 있다.

$$
D=\{x_i\}_{i=1}^{n}
$$

주어진 Data 안에서 의미 있는 Structure나 Pattern을 발견하는 것이 목적이다.

대표적인 문제는 다음과 같다.

- Clustering
- Dimensionality Reduction
- Anomaly Detection

예를 들어 기업이 많은 Customer의 구매 Data를 가지고 있다고 하자.

미리 정해진 정답 없이 구매 행동이 비슷한 Customer들을 여러 Group으로 나누고 싶을 수 있다.

Machine Learning Algorithm은 Data 자체에서 이러한 Structure를 찾는다.


### Reinforcement Learning

**Reinforcement Learning**에서는 **Agent**가 **Environment**와 상호작용하면서 학습한다.

Agent는 현재 **State**를 관찰하고 **Action**을 선택한다.

그 결과로 Environment로부터 **Reward**를 받는다.

간단하게 표현하면 다음과 같다.

> **State → Action → Reward**

Agent는 장기적으로 더 큰 Reward를 얻을 수 있는 Action을 학습한다.

예를 들어 Maze에서 Goal까지 이동하는 Agent를 생각할 수 있다.

좋은 Action에는 Positive Reward를 주고 잘못된 Action에는 Negative Reward를 줄 수 있다.

Agent는 이러한 경험을 통해 적절한 행동, 즉 **Policy**를 학습한다.

## AI, Machine Learning, and Deep Learning

**Artificial Intelligence**, **Machine Learning**, **Deep Learning**은 서로 관련되어 있지만 같은 개념은 아니다.


### Artificial Intelligence

**Artificial Intelligence (AI)**가 가장 넓은 개념이다.

AI의 목적은 Computer가 지능적인 행동을 필요로 하는 Task를 수행하도록 만드는 것이다.

예를 들면 다음과 같다.

- Reasoning
- Planning
- Problem Solving
- Language Understanding
- Image Recognition
- Decision Making


### Machine Learning

**Machine Learning (ML)**은 AI를 구현하기 위한 하나의 방법이다.

모든 Rule을 사람이 직접 Programming하는 대신 Data로부터 Pattern과 Relationship을 학습한다.

따라서

$$
ML \subset AI
$$

이다.


### Deep Learning

**Deep Learning (DL)**은 여러 Layer를 가진 **Artificial Neural Network**를 주로 이용하는 Machine Learning의 한 분야이다.

따라서

$$
DL \subset ML
$$

이다.

전체적인 관계는 다음과 같다.

$$
DL \subset ML \subset AI
$$

또는

$$
AI \supset ML \supset DL
$$

로 표현할 수 있다.

따라서 다음이 성립한다.

- Deep Learning은 Machine Learning이다.
- Machine Learning은 Artificial Intelligence의 일부이다.
- 모든 Machine Learning이 Deep Learning인 것은 아니다.
- 모든 Artificial Intelligence가 Machine Learning인 것은 아니다.

## How We Will Study Machine Learning

이 강의에서는 다양한 Machine Learning Method를 배운다.

각 Method는 가능한 한 다음과 같은 과정으로 학습한다.

> **Problem → Problem Analysis → Method → Application → Result Analysis**

각 Method를 배울 때 다음을 생각한다.

- 어떤 Problem을 해결하려고 하는가?
- 어떤 Data를 가지고 있는가?
- 어떤 Method를 사용할 수 있는가?
- Method를 어떻게 적용하는가?
- Result는 무엇을 의미하는가?
- Result는 합리적인가?

Method를 이해하고 사용하는 데 필요한 Theory는 함께 학습한다.

그러나 이 강의의 목적은 Formula나 Algorithm을 단순히 암기하는 것이 아니다.

목표는

> **Data를 이용하여 Problem을 해결하기에 적절한 Method를 선택하고 적용하는 능력**

을 기르는 것이다.

> 어떤 Data를 사용하고 있는가?

> 각 Variable은 무엇을 의미하는가?

> 왜 이 Machine Learning Method를 선택했는가?

> AI가 생성한 Code가 실제로 내가 원하는 일을 하는가?

> Result가 합리적인가?

AI는 **Code**를 만드는 데 도움을 줄 수 있다.

그러나 **Problem을 정의하고, 판단하고, Result를 해석하는 것은 사람의 역할**이다.

따라서 이 강의에서는 AI를 **Problem Solving을 위한 Tool**로 사용한다.
