# Machine Learning Problems and Data

이번 강의에서는 현실의 Problem을 Machine Learning이 처리할 수 있는 Data Problem으로 표현하는 방법을 배운다.

Machine Learning은 Algorithm에서 시작하지 않는다.

먼저 해결하고자 하는 Problem이 있고, 그 Problem과 관련된 Data가 있다.

예를 들어 다음과 같은 Problem을 생각해 보자.

> 학생의 공부시간, 출석률, 과제점수 등의 Data를 이용하여 Final Examination의 점수를 예측할 수 있을까?

이 단계에서는 아직 어떤 Machine Learning Method를 사용할지 결정하지 않는다.

먼저 이 Problem을 Machine Learning이 처리할 수 있는 형태로 표현해야 한다.

> Problem → Data → Input / Output → Model → Learning → Result

## Problem

가장 먼저 해야 할 일은 무엇을 하고자 하는지를 명확하게 정의하는 것이다.

예를 들어 다음 질문들은 서로 다른 Problem이다.

- 주택의 판매가격을 예측하고 싶다.
- Email이 Spam인지 판단하고 싶다.
- 비슷한 구매행동을 보이는 Customer들을 찾고 싶다.
- Game에서 높은 Score를 얻는 행동을 학습하고 싶다.

Problem이 달라지면 필요한 Data도 달라지고 사용할 Machine Learning Method도 달라진다.

따라서 Machine Learning의 출발점은 다음 질문이다.

> 무엇을 하고자 하는가?

## Data

Problem을 정의했다면 다음으로 생각할 것은 Data이다.

> Problem을 해결하기 위해 어떤 Data를 사용할 수 있는가?

학생의 Final Examination Score를 예측한다고 하자.

다음과 같은 Data가 있을 수 있다.

| Study Hours | Attendance | Assignment Score | Final Score |
|---:|---:|---:|---:|
| 5 | 70 | 65 | 62 |
| 8 | 80 | 72 | 70 |
| 12 | 90 | 85 | 84 |
| 15 | 95 | 92 | 91 |

Data를 가지고 있다는 것만으로 Machine Learning Problem이 완성되는 것은 아니다.

각 Data가 Problem에서 어떤 역할을 하는지 결정해야 한다.

## Input and Output

Machine Learning Problem에서는 Data를 Input과 Output의 관점에서 생각하는 것이 중요하다.

Input을 일반적으로 $X$, Output을 $y$로 표현한다.

$$
X \rightarrow y
$$

학생 성적 예측 Problem에서는 다음과 같이 생각할 수 있다.

$$
X = \{\text{Study Hours, Attendance, Assignment Score}\}
$$

$$
y = \text{Final Score}
$$

따라서 우리가 찾고자 하는 관계는

$$
y=f(X)
$$

이다.

## Feature

Input으로 사용하는 각각의 정보를 Feature라고 한다.

학생 성적 예측 Problem에서는 다음이 Feature가 될 수 있다.

- Study Hours
- Attendance
- Assignment Score

한 학생의 Input은 다음과 같이 표현할 수 있다.

$$
x_i=(x_{i1},x_{i2},x_{i3})
$$

예를 들어 어떤 학생의 Data가

- Study Hours = 12
- Attendance = 90
- Assignment Score = 85

라면,

$$
x_i=(12,90,85)
$$

가 된다.

## Dataset as $X$ and $y$

여러 학생의 Feature를 모으면 Input Data $X$를 만들 수 있다.

$$
X=
\begin{bmatrix}
5 & 70 & 65 \\
8 & 80 & 72 \\
12 & 90 & 85 \\
15 & 95 & 92
\end{bmatrix}
$$

이에 대응하는 Final Score는

$$
y=
\begin{bmatrix}
62 \\
70 \\
84 \\
91
\end{bmatrix}
$$

로 표현할 수 있다.

Machine Learning에서는 Dataset을 단순한 Table로 보는 것에서 한 단계 더 나아가 $X$와 $y$의 관계로 이해하는 것이 중요하다.

## Model

우리가 알고 싶은 실제 관계를 다음과 같이 나타내자.

$$
y=f(X)
$$

그러나 실제 Problem에서는 일반적으로 $f$를 알지 못한다.

Machine Learning에서는 Data를 이용하여 $f$를 근사하는 Model을 찾는다.

$$
\hat{y}=\hat{f}(X)
$$

여기서

- $X$ : Input Data
- $y$ : 실제 Output
- $\hat{f}$ : 학습된 Model
- $\hat{y}$ : Model이 예측한 Output

이다.

Model은 Input으로부터 Output을 만들어 내는 관계를 나타낸다.

## A Simple Model

학생 성적을 예측하는 매우 단순한 Model을 다음과 같이 생각할 수 있다.

$$
\hat{y}=w_1x_1+w_2x_2+w_3x_3+b
$$

여기서

- $x_1$ : Study Hours
- $x_2$ : Attendance
- $x_3$ : Assignment Score

라고 하자.

$w_1$, $w_2$, $w_3$, $b$의 값을 알게 되면 새로운 학생의 정보를 이용하여 Final Score를 예측할 수 있다.

그렇다면 다음 Problem은 이 값들을 어떻게 결정할 것인가이다.

## Learning

Model의 형태를 정했다고 하더라도 아직 문제가 해결된 것은 아니다.

$$
\hat{y}=w_1x_1+w_2x_2+w_3x_3+b
$$

라는 Model을 사용한다면 다음 값을 결정해야 한다.

$$
w_1,\ w_2,\ w_3,\ b
$$

이 값들을 사람이 임의로 정하는 것이 아니라 Data를 이용하여 결정한다.

이 과정을 Learning 또는 Training이라고 한다.

> Data를 이용하여 Model이 좋은 결과를 만들도록 Model의 Parameter를 결정하는 과정

을 Learning이라고 할 수 있다.

> Data + Model → Learning → Trained Model

## Parameter

Learning을 통해 결정되는 Model 내부의 값을 Parameter라고 한다.

예를 들어

$$
\hat{y}=w_1x_1+w_2x_2+w_3x_3+b
$$

에서는

$$
w_1,\ w_2,\ w_3,\ b
$$

가 Parameter이다.

Learning의 목적은 주어진 Data를 잘 설명할 수 있는 Parameter를 찾는 것이다.

## Prediction and Error

Model이 예측한 값 $\hat{y}$와 실제 값 $y$는 일반적으로 완전히 같지 않다.

예를 들어 실제 Final Score가

$$
y=84
$$

인데 Model이

$$
\hat{y}=80
$$

이라고 예측했다고 하자.

실제 값과 예측 값 사이에는 Error가 존재한다.

$$
Error=y-\hat{y}
$$

Learning에서는 이러한 Error가 작아지도록 Parameter를 조정한다.

Machine Learning의 기본적인 과정을 간단히 표현하면 다음과 같다.

> Data → Model → Prediction → Error → Parameter Update

## Supervised Learning

Supervised Learning에서는 Input $X$와 Output $y$가 함께 주어진다.

$$
D=\{(x_i,y_i)\}_{i=1}^{n}
$$

목적은 Data를 이용하여

$$
X \rightarrow y
$$

의 관계를 학습하는 것이다.

대표적인 Supervised Learning Problem은 다음과 같다.

- Regression
- Classification

### Regression

Regression은 Numerical Value를 예측하는 Problem이다.

예를 들어 다음 Problem을 생각해 보자.

> 주택 정보를 이용하여 주택 가격을 예측하고 싶다.

Input은

$$
X=\{\text{Size, Number of Rooms, Age}\}
$$

이고 Output은

$$
y=\text{Price}
$$

이다.

따라서

$$
\text{House Information} \rightarrow \text{Price}
$$

의 관계를 Data로부터 학습한다.

Output이 Numerical Value이므로 Regression Problem이다.

### Regression

Regression은 Numerical Value를 예측하는 Problem이다.

예를 들어 다음 Problem을 생각해 보자.

> 주택 정보를 이용하여 주택 가격을 예측하고 싶다.

Input은

$$

## Unsupervised Learning

Unsupervised Learning에서는 Input Data $X$는 있지만 정답에 해당하는 $y$가 주어지지 않는다.

$$
D=\{x_i\}_{i=1}^{n}
$$

예를 들어 Online Shopping Mall이 다음과 같은 Customer Data를 가지고 있다고 하자.

- Age
- Number of Purchases
- Average Purchase Amount
- Visit Frequency

회사는 구매행동이 비슷한 Customer들을 찾아 Group으로 나누고 싶다.

이 경우

$$
X=\text{Customer Data}
$$

는 있지만 미리 정해진 $y$는 없다.

목적은 Data 안에 존재하는 Pattern이나 Structure를 발견하는 것이다.

Clustering은 대표적인 Unsupervised Learning Method이다.

## Reinforcement Learning

Reinforcement Learning에서는 Agent가 Environment와 상호작용하면서 학습한다.

Agent는 State $s$에서 Action $a$를 선택한다.

그 결과 Reward $r$을 받고 새로운 State $s'$로 이동한다.

하나의 Experience는 다음과 같이 표현할 수 있다.

$$
(s,a,r,s')
$$

Agent는 이러한 Experience를 반복하면서 어떤 State에서 어떤 Action을 선택해야 하는지를 학습한다.

$$
\pi:s\rightarrow a
$$

여기서 $\pi$를 Policy라고 한다.

예를 들어 Maze에서 Goal을 찾아가는 Agent를 생각할 수 있다.

- State : Agent의 현재 위치
- Action : 이동 방향
- Reward : 이동 결과에 따라 받은 값

목적은 높은 Reward를 얻을 수 있는 Policy를 학습하는 것이다.

## The Same Data Can Produce Different Problems

같은 Data를 가지고 있더라도 무엇을 하고자 하는가에 따라 서로 다른 Machine Learning Problem을 만들 수 있다.

다음과 같은 Student Data가 있다고 하자.

| Study Hours | Attendance | Assignment | Final Score |
|---:|---:|---:|---:|
| 5 | 70 | 65 | 62 |
| 8 | 80 | 72 | 70 |
| 12 | 90 | 85 | 84 |
| 15 | 95 | 92 | 91 |

이 Data로 다음과 같은 서로 다른 Problem을 만들 수 있다.

- Study Hours를 이용하여 Final Score를 예측한다.
- Study Hours, Attendance, Assignment를 이용하여 Final Score를 예측한다.
- 학생이 Pass할 것인지 Fail할 것인지 예측한다.
- 학습행동이 비슷한 학생들을 Group으로 나눈다.

따라서 같은 Data라도 Problem을 어떻게 정의하느냐에 따라 $X$, $y$, 그리고 사용할 Method가 달라질 수 있다.

> Data가 Problem을 결정하는 것이 아니라, 하고자 하는 일이 Problem을 결정한다.

## From Problem to Machine Learning

새로운 Problem을 만나면 다음 질문을 순서대로 생각한다.

### What do we want to do?

해결하고자 하는 일을 한 문장으로 표현한다.

> 자동차의 정보를 이용하여 Fuel Efficiency를 예측하고 싶다.

### What data do we have?

사용할 수 있는 Data를 확인한다.

- Horsepower
- Weight
- Engine Size
- Fuel Efficiency

### What is $X$?

$$
X=\{\text{Horsepower, Weight, Engine Size}\}
$$

### What is $y$?

$$
y=\text{Fuel Efficiency}
$$

### What type of Problem is it?

$X$와 $y$가 모두 존재하므로 Supervised Learning이다.

$y$가 Numerical Value이므로 Regression Problem이다.

### What should be learned?

Data를 이용하여

$$
\hat{y}=\hat{f}(X)
$$

를 만족하는 Model을 학습한다.
