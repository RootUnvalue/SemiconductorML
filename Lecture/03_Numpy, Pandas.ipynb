## NumPy

NumPy는 Numerical Data를 효율적으로 처리하기 위한 Python Library이다.

NumPy에서 가장 기본적인 Data Structure는 ndarray이다.

Python의 List와 비슷하게 여러 값을 저장할 수 있지만, Numerical Calculation을 효율적으로 수행할 수 있다.

일반적으로 NumPy를 다음과 같이 불러온다.

```python
import numpy as np

data = [10, 20, 30, 40, 50]

x = np.array(data)

print(x)
print(type(x))
```

```python
print(data)
print(type(data))
```

NumPy Array의 중요한 특징 중 하나는 Array 전체에 Mathematical Operation을 쉽게 적용할 수 있다는 것이다.

예를 들어 모든 값에 2를 곱하려고 한다.

```python
print(x * 2)
```

```python
print(data * 2)
```

Python List에 `* 2`를 적용하는 것과 NumPy Array에 `* 2`를 적용하는 것은 의미가 다르다.

NumPy에서는 각 Element에 Mathematical Operation이 적용된다.

Machine Learning에서는 많은 Numerical Data에 동일한 계산을 수행해야 하므로 이러한 기능이 매우 유용하다.

## Two-Dimensional Array

Machine Learning Data는 하나의 값만 가지는 경우보다 여러 Feature를 가지는 경우가 많다.

예를 들어 학생 한 명에 대해 다음 세 가지 Feature를 사용한다고 하자.

- Study Hours
- Attendance
- Assignment Score

네 명의 학생에 대한 Input Data $X$는 다음과 같이 표현할 수 있다.

$$
X=
\begin{bmatrix}
5 & 70 & 65 \\
8 & 80 & 72 \\
12 & 90 & 85 \\
15 & 95 & 92
\end{bmatrix}
$$

이를 NumPy의 Two-Dimensional Array로 만들 수 있다.

```python
X = np.array([
    [5, 70, 65],
    [8, 80, 72],
    [12, 90, 85],
    [15, 95, 92]
])

print(X)
print(X.shape)
```

## Selecting Data

NumPy Array에서는 필요한 Sample이나 Feature를 선택할 수 있다.

Machine Learning에서는 전체 Dataset을 항상 그대로 사용하는 것이 아니라 필요한 Row와 Column을 선택해야 하는 경우가 많다.

Python의 Index는 0부터 시작한다.

```python
# 첫 번째 Sample 선택
print(X[0])

# 모든 Sample의 첫 번째 Feature 선택
study_hours = X[:, 0]
print(study_hours)

# 첫 번째와 두 번째 Feature 선택
features = X[:, 0:2]    # slicing   a:b  a<=x<b
print(features)
```

## From NumPy to Pandas

NumPy는 Numerical Calculation에 매우 유용하다.

그러나 실제 Dataset에서는 각 Column이 무엇을 의미하는지 아는 것이 중요하다.

예를 들어 다음과 같은 Data를 생각해 보자.

- Country
- Year
- Population
- Income
- Life Expectancy

이처럼 Row와 Column으로 구성되고 각 Column이 의미를 가지는 Data를 다룰 때는 Pandas의 DataFrame이 편리하다.

```python
import pandas as pd

# Dictionary를 이용하여 DataFrame 생성
data = {
    "StudyHours": [5, 8, 12, 15],
    "Attendance": [70, 80, 90, 95],
    "Assignment": [65, 72, 85, 92],
    "FinalScore": [62, 70, 84, 91]
}

df = pd.DataFrame(data)

# DataFrame 확인
df
```

## Reading and Exploring CSV Data

실제 Machine Learning Data는 CSV File로 제공되는 경우가 많다.

Pandas의 `read_csv()`를 이용하면 CSV File을 DataFrame으로 읽을 수 있다.

CSV File을 읽은 후에는 바로 Machine Learning을 적용하기보다 먼저 Data의 구조와 내용을 확인해야 한다.

주로 다음을 확인한다.

- Data의 일부

- Column Name
- Data Type

```python
# CSV File을 DataFrame으로 읽기
df = pd.read_csv(data_path+"life_expectancy.csv")

# 처음 5개의 Sample 확인
display(df.head())

# Data의 크기 확인
print("Shape:", df.shape)

# Column Name 확인
print("Columns:", df.columns.tolist())

# 각 Column의 Data Type 확인
print("\nData Types:")
print(df.dtypes)
```

## Selecting Data

DataFrame에서 Column Name을 이용하여 필요한 Variable을 선택할 수 있다.

하나의 Column을 선택할 수도 있고 여러 Column을 동시에 선택할 수도 있다.

또한 조건을 이용하여 특정 Sample만 선택할 수 있다.

어떤 Data를 선택할 것인지는 해결하고자 하는 Problem에 따라 결정한다.

```python
# 하나의 Column 선택
life = df["Life expectancy"]
print(life.head())

# 여러 Column 선택
selected = df[["Income composition of resources", "Population", "Life expectancy"]]
display(selected.head())

# 특정 조건을 만족하는 Sample 선택
high_income = df[df["Income composition of resources"] > 0.5]
display(high_income.head())
```

## Transforming Data

기존 Variable을 그대로 사용하는 것뿐 아니라 Problem에 맞게 새로운 Variable을 만들 수도 있다.

이러한 과정을 Data Transformation이라고 한다.

예를 들어 단위를 변경하거나 기존 Variable을 이용하여 새로운 Feature를 만들 수 있다.

```python
# Income의 단위를 1000 단위로 변환하여 새로운 Column 생성
df["Income1000"] = df["Income composition of resources"] / 1000

# 변환 결과 확인
display(df.head())
```

## From DataFrame to $X$ and $y$

Machine Learning Problem을 다음과 같이 정의했다고 하자.

> 국가의 여러 정보를 이용하여 Life Expectancy를 예측하고 싶다.

이 Problem에서 예측에 사용할 Variable을 $X$, 예측하고자 하는 Variable을 $y$로 선택한다.

$$
X \rightarrow y
$$

중요한 것은 DataFrame 전체가 자동으로 $X$가 되는 것이 아니라는 점이다.

$X$와 $y$는 해결하려는 Problem에 따라 우리가 결정한다.

```python
# Input X 선택
X = df[["Income composition of resources", "Population", "Schooling"]]

# Output y 선택
y = df["Life expectancy"]

# 선택한 X와 y 확인
display(X.head())
display(y.head())

# Shape 확인
print("X shape:", X.shape)
print("y shape:", y.shape)
```
